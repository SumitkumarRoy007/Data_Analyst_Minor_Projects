```python
#import libraries
import pandas as pd
import numpy as np
```

```python
#import database 
df=pd.read_csv("Emp_data.csv")
df.head()
```

```python
# 1.check info
df.info()
```

```python
#2. Department wise count
df['department'].value_counts()
```

```python
#3. attrition_status count
df['attrition_status'].value_counts()
```

```python
#4. Average salary by department
df.groupby('department')['salary_lpa'].agg(['mean','min','max','count']).round(2)
```

```python
#5. create new column= "exact_salary"
df['exact_salary']=(df['salary_lpa']*100000).astype(int)
#6. create a new column name of "monthly salary"
df['monthly_salary']=(df['exact_salary']/12).astype(int).round(2)
df[['employee_name','department','total_experience','age','monthly_salary']].head(10)
```

```python
#filling null values
# For attrition status fill with not active
df['attrtion_status']=df['attrition_status'].fillna('Not Active')
df['attrtion_status']
```

```python
#filling null values
# for numeric columns
df['projects_handled']=df['projects_handled'].fillna(df['projects_handled']).median()
df['projects_handled']
```

```python
#filling the null values with NA
df['attrition_reason']=df['attrition_reason'].fillna('N/A - Active')
df['attrition_reason']
```

```python
# Education wise employee count
df.groupby('education')['education'].value_counts().sort_values()
```

```python
# Maratial status wise employee count
df.groupby('marital_status')['department'].value_counts()
```

```python
# Department wise average monthly salary
df.groupby('department')['monthly_salary'].mean().round(0).sort_values()
```

```python
# Saving the file
df.to_csv("srijan_emp.csv",index = False)
print("File Saved!")
```
